### List

```python
age_list=[24,25,26,27,28]
age_list
```

```python
name_list=["Deenu","Manju","Sara","Kiran","Balu"]
name_list
```

```python
name_list.append("Yazhini") 
name_list
```

```python
age_list.insert(2, 30)
age_list
```

```python
name_list.remove("Yazhini")
name_list
```

```python
age_list.pop()
age_list
```

```python
age_list.extend([29, 30, 26])
age_list
```

```python
age_list.sort(reverse=True) 
age_list
```

```python
print("Max:", max(age_list))               
print("Min:", min(age_list))
print("Sum:", sum(age_list))
```

```python
print(name_list[0])                         
print(name_list[-1])                        
print(name_list[2:5])                       
print(name_list[::-1]) 
```

### Dictionary

```python
student_marks = {
    "Kiran": 75,
    "Manju": 60,
    "Deenu": 90,
    "Sara": 85,
    "Balu": 70}
student_marks
```

```python
print("Kiran's mark:", student_marks["Kiran"])
```

```python
student_marks["Janani"] = 80
student_marks
```

```python
student_marks["Manju"] = 82
student_marks
```

```python
print("Keys:", student_marks.keys())
print("Values:", student_marks.values())
print("Items:", student_marks.items())
```

### Set

```python
my_set = {'a', 'e', 'i', 'o', 'u', 'a', 'a', 'i'}
print("my_set:", my_set)
# Explanation: Sets remove duplicates, so repeated 'a' and 'i' appear only once.
```

```python
# my_set[4] = 's' → ERROR
# Explanation: Sets are unordered and do not support indexing or item assignment.
```

```python
set1 = {1, 3, 5, 7, 9}
set2 = {2, 3, 5, 8, 10}
print (set1)
print (set2)
```

```python
print("Union:", set1.union(set2))
print("Intersection:", set1.intersection(set2))
```

### Operators & Conditional Statements

```python
score = int(input("Enter your score (0 to 10): "))

if score > 7:
    print("Above Average: Excellent work! Keep it up.")
elif 4 <= score <= 7:
    print("Average: Good effort! Keep practicing, there's room for improvement.")
else:
    print("Below Average: Need to improve your performance, consistent practice will lead to better results.")
```
